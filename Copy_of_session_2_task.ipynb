````python
#Ai's code

import math

x = 0.0
y = 0.0

while True:
    op_str = input("Enter operation (e.g., 'UP 5', '!', etc.): ")

    if op_str == '!':
        break

    parts = op_str.split()
    if len(parts) != 2:
        print("Invalid input format. Please use 'DIRECTION STEPS' or '!' to stop.")
        continue

    direction = parts[0].upper()
    try:
        value = float(parts[1])
    except ValueError:
        print("Invalid step value. Please enter a number.")
        continue

    if direction == "UP":
        y += value
    elif direction == "DOWN":
        y -= value
    elif direction == "LEFT":
        x -= value
    elif direction == "RIGHT":
        x += value
    else:
        print("Invalid direction. Use UP, DOWN, LEFT, or RIGHT.")

# Calculate the Euclidean distance from the origin (0,0)
distance = math.sqrt(x**2 + y**2)

# Print the nearest integer
print(int(round(distance)))
````
